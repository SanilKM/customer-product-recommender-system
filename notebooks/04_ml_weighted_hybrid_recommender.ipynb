# 04 - ML-Weighted Hybrid Recommender

This notebook trains a supervised ML model using July 2026 conversion data to learn signal weights for the hybrid recommender.

## Goal

Instead of manually choosing weights for propensity, digital intent, transaction interest, profile fit, segment affinity, and co-occurrence, we train a model to learn which signals are useful for next-month conversion.

```
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

model_input_dir = PROJECT_ROOT / 'data' / '03_model_outputs'
conversion_data = PROJECT_ROOT / 'data' / '01_raw' / 'conversion_data_july_2026.csv'
output_dir = PROJECT_ROOT / 'data' / '04_ml_weighted_hybrid_outputs'
script_path = PROJECT_ROOT / 'src' / 'ml_weighted_hybrid_recommender.py'

print('Project root:', PROJECT_ROOT)
print('Model input folder exists:', model_input_dir.exists(), model_input_dir)
print('Conversion data exists:', conversion_data.exists(), conversion_data)
print('Script exists:', script_path.exists(), script_path)
```

```
import subprocess
import sys

output_dir.mkdir(parents=True, exist_ok=True)

cmd = [
    sys.executable,
    str(script_path),
    '--model-input-dir', str(model_input_dir),
    '--conversion-data', str(conversion_data),
    '--output-dir', str(output_dir),
]

result = subprocess.run(cmd, capture_output=True, text=True)
print(result.stdout)
if result.stderr:
    print(result.stderr)
```

```
import pandas as pd

weights = pd.read_csv(output_dir / 'ml_learned_signal_weights.csv')
metrics = pd.read_csv(output_dir / 'ml_model_test_metrics.csv')
comparison = pd.read_csv(output_dir / 'ml_weighted_hybrid_vs_baselines_test_accuracy.csv')
sanity = pd.read_csv(output_dir / 'ml_weighted_hybrid_sanity_checks.csv')

print('ML learned signal weights')
display(weights)

print('ML test metrics')
display(metrics)

print('Top-N accuracy comparison on test customers')
display(comparison)

print('Sanity checks')
display(sanity)
```
